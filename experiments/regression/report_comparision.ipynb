```python
import os 
from itables import show
import pandas as pd
import plotly.express as px
from plotly.offline import init_notebook_mode
init_notebook_mode(connected=True)
%matplotlib inline

import Orange #version 3.33
from Orange.evaluation import compute_CD, graph_ranks
```

```python
def plot_box(df, for_field, title, y_axis):
    fig = px.bar(df, x='dataset', color="model",
                y=for_field,
                barmode='group',
                #text_auto=True,
                #height=560,
                #width=3000,
                title=title,
                )
    fig.update_traces(textposition='outside')
    fig.update_layout(
                title={
                'x':0.5,
                'xanchor': 'center'
            })
    fig.update_layout(yaxis_title=y_axis)
    fig.show()
```

```python
results_df = pd.read_csv(f"./results/comparision/exceptions_summary_ALL.csv")
```

```python
show(results_df)
```

### Wykresy

#### RMSE on test

```python
plot_box(results_df, "rmse_test", title='RMSE on test', y_axis="RMSE")
```

#### Liczba_regul

```python
plot_box(results_df, "liczba_regul", title='Liczba regul', y_axis="Liczba regul")
```

### Tabele

```python
def get_table(df, wskaznik):
    df = results_df[["dataset", "model", wskaznik]]

    df = df.pivot(
        index="dataset", 
        columns= "model", 
        values=wskaznik
        )

    show(df)
```

#### RMSE on test

```python
wskaznik = "rmse_test"
get_table(results_df, wskaznik)
```

#### Liczba_regul

```python
wskaznik = "liczba_regul"
get_table(results_df, wskaznik)
```

### Ranking & CD_diagramy

```python
import matplotlib.pyplot as plt
```

#### Zbiór testowy

```python
ranking_df = results_df
ranking_df = ranking_df.pivot(index='dataset', columns='model', values='rmse_test')
ranking_df.columns.name = None

ranked_df = ranking_df.rank(axis=1, method='average', ascending=True)
```

```python
average_ranks = ranked_df.mean()
average_ranks
```

```python
plt.figure(figsize=(10, 4))

cd = compute_CD(alpha="0.05", avranks=average_ranks, n = ranked_df.shape[0],  test = "nemenyi")

plot = graph_ranks(avranks=average_ranks, names = ranked_df.columns.values, cd = cd)

plt.savefig("./plots/rmse_test.pdf", bbox_inches="tight")  # wektorowy!
plt.savefig("./plots/rmse_test.png", dpi=600, bbox_inches="tight")  # raster high-res
plt.close()
```
